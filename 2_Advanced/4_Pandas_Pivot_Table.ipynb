```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

##Counts of Job Titles

```python
df.pivot_table(index='job_title_short',aggfunc='size')
```

```python
df.groupby('job_title_short').size()
```

```python
df.pivot_table(index='job_title_short', values='salary_year_avg', aggfunc='median')
```

```python
df.groupby('job_title_short')['salary_year_avg'].median()
```

```python
df.pivot_table(index='job_title_short', values='salary_year_avg', columns='job_country', aggfunc='median')
```

```python
df.pivot_table(index='job_country', values='salary_year_avg', columns='job_title_short', aggfunc='median')
```

```python
top_countries=df['job_country'].value_counts().head()
top_countries
```

```python
top_countries=df['job_country'].value_counts().head(6).index

df_job_country_salary = df.pivot_table(index='job_country', values='salary_year_avg', columns='job_title_short', aggfunc='median')
```

```python
df_job_country_salary = df_job_country_salary.loc[top_countries]
df_job_country_salary
```

```python
job_titles = ['Data Analyst','Data Engineer','Data Scientist']
df_job_country_salary=df_job_country_salary[job_titles]
df_job_country_salary
```

```python
df_job_country_salary.plot(kind='bar')
plt.ylabel('Median Salary($USD)')
plt.xlabel('')
plt.title('Median Salary by Country and Job Title')
plt.xticks(rotation=45,ha='right')
```
