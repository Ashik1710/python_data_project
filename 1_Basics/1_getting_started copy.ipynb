```python
for job in job_roles:          # loop through each job dictionary
    qualified = True           # assume you're qualified at the start
    for skill in my_skills:    # loop through each of your skills
        if skill not in job['skills']:   # check if this skill is missing
            qualified = False            # mark as not qualified
            print(qualified)             # print False
            print(skill)                 # print the missing skill
            print(job['skills'])         # print the job's required skills
            break                        # stop checking further skills
```

```python
for job in job_roles:          # loop through each job dictionary
    qualified = True           # assume you're qualified at the start
    for skill in my_skills:    # loop through each of your skills
        if skill not in job['skills']:   # check if this skill is missing
            qualified = False            # mark as not qualifie
            break
        if qualified:
          print(job)                      # stop checking further skills
```

```python
for job in job_roles:
    qualified = True
    for skill in my_skills:
        if skill not in job['skills']:
            qualified = False
            break
    if qualified:
        print(job)   # print job only once if all your skills are present
```

```python
# Install datasets Library (if not already installed)
# !pip install datasets

# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
job_salary=df.groupby('job_title_short')['salary_year_avg'].median().sort_values()
```

```python
job_salary.plot(kind='bar')
plt.title('Job Salary')
plt.xlabel('Job Title')
plt.ylabel('Salary')
```
